## Queue

```
from prodlab.reliability import ReviewQueue

queue = ReviewQueue()
queue.add("customer thumbs down", "Refund answer was unclear", 1.0)
queue.add("low judge score", "Reply missed the second question", 2.0)
queue.add("safety block", "Planted instruction in returns policy", 3.0)
queue.add("upset customer", "Parcel late for the third time", 4.0)
queue.add("safety block", "Asked for another customer's order", 5.0)
while (item := queue.next()) is not None:     # most urgent first
    print(f"{item.reason:<21} {item.text}")
    queue.decide(item, ok=item.reason != "low judge score")
print("new labelled examples for the judge:", len(queue.decisions))
```

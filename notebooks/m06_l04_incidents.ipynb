## Incident

```
from prodlab.reliability import incident_timeline

rows, events = incident_timeline()          # simulated, seeded
for name in ("release", "alert", "rollback"):
    minute = events[name]
    rate = rows[minute][1]
    print(f"minute {minute:2d}: {name:<8} error rate {rate:4.1f}%")
print("time to detect:  ", events["alert"] - events["release"], "min")
print("time to mitigate:", events["rollback"] - events["alert"], "min")
```

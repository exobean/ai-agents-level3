## Calibrate

```
from prodlab.evals import LABELLED, calibrate, passes

for version in (1, 2, 3):
    r = calibrate(version)
    print(f"judge v{version}: agreement {r['agreement']:.0%}, "
          f"kappa {r['kappa']:.2f}")
for question, answer, human in LABELLED:
    if passes(question, answer, 3) != human:
        print("v3 still disagrees on:", answer[:40], "...")
```

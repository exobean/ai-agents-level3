## Gate

```
from agentlab.assistant import ShopAssist2
from agentlab.scorecard import GATES, run_cases, scores
from prodlab.evals import release_check

live = scores(run_cases())                         # the running release
candidate = scores(run_cases(lambda: ShopAssist2(budget=2)))
for name in live:
    print(f"{name:<16} live {live[name]:<4}", end="  ")
    print(f"candidate {candidate[name]}")
verdict = release_check(live, candidate, GATES)
print("regressions:", verdict["regressions"])
print("failing gates:", verdict["failing_gates"])
print("ship it?", verdict["ship"])
```

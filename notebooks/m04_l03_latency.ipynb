## Latency

```
from prodlab.perf import latency

for message in [
    "Where is my order #1042?",
    "Where is my order #1043? And also, is there a student discount?",
    "Where is my order #1042? And also, is there a student discount? "
    "Can I pay cash on delivery?",
]:
    seq, par = latency(message, False), latency(message, True)
    print(f"{seq:6.0f} ms in turn  {par:6.0f} ms in parallel  "
          f"({message.count('?')} part(s))")
```

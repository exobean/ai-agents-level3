## Metrics

```
from agentlab.assistant import ShopAssist2
from prodlab.obs import percentile, simulate_traffic

rows = simulate_traffic(ShopAssist2, n=400)     # seeded, repeatable
latency = [r["latency_ms"] for r in rows]
errors = sum(r["error"] for r in rows)
print(f"requests: {len(rows)}   errors: {errors}")
print(f"mean: {sum(latency) / len(latency):6.0f} ms")
for p in (50, 95, 99):
    print(f"p{p}:  {percentile(latency, p):6.0f} ms")
print(f"model calls per request: "
      f"{sum(r['calls'] for r in rows) / len(rows):.1f}")
```

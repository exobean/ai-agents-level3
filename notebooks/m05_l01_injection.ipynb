## Injection

```
import textwrap

from agentlab.assistant import ShopAssist2
from prodlab.security import SecureAssistant, poisoned_index, scan

question = "How long does a refund take?"
with poisoned_index():           # an attacker edited a policy document
    plain = ShopAssist2().ask("cust-7", question)["reply"]
    guarded = SecureAssistant(layers=("retrieval",)).ask(
        "cust-7", question)["reply"]
for name, reply in [("plain", plain), ("retrieval filter", guarded)]:
    print(f"{name}:")
    print(textwrap.fill(reply, 66, initial_indent="   ",
                        subsequent_indent="   "))
print("scan of the plain reply:")
print("  ", scan(plain))
```

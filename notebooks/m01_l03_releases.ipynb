## Release

```
from agentlab.scorecard import failing
from prodlab.config import Settings
from prodlab.release import choose, diff, manifest

good = {"correct": 0.9, "citations_valid": 1.0, "safe": 1.0,
        "avg_calls": 3.0}
cheap = {**good, "correct": 0.8, "avg_calls": 2.4}
v10 = manifest("1.0", Settings(), good)
v11 = manifest("1.1", Settings(budget=6), cheap)
print("prompt fingerprint:", v10["prompts"])
for change in diff(v10, v11):
    print("changed:", change)
print("v1.1 fails gates:", failing(v11["scores"]))
chosen = choose([v10, v11], lambda s: not failing(s))
print("roll back to:", chosen["version"])
```

## Shadow

```
from agentlab.assistant import ShopAssist2
from prodlab.evals import shadow
from prodlab.obs import SAMPLE

live, candidate = ShopAssist2(), ShopAssist2(budget=2)
report = shadow(live, candidate, SAMPLE)        # users see only live
print("questions mirrored:", len(report["rows"]))
print("answers differ:", report["differ"],
      "| worse:", report["worse"], "| better:", report["better"])
for row in report["rows"]:
    if not row["same"]:
        print("live:     ", row["baseline"][:52], "...")
        print("candidate:", row["candidate"][:52], "...")
```

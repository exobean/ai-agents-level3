## Cache

```
from prodlab.perf import CachedAssistant

safe = CachedAssistant()                     # personal answers per user
for user in ["cust-a", "cust-b", "cust-c"]:
    out = safe.ask(user, "How long does a refund take?")
    print(user, "cached" if out["cached"] else "computed")

unsafe = CachedAssistant(safe=False)         # one key for everyone
unsafe.ask("cust-a", "Hi, my order number is #1043")
unsafe.ask("cust-a", "Where is it?")
print("cust-b asks 'Where is it?' and gets:")
print("  ", unsafe.ask("cust-b", "Where is it?")["reply"])
```

## Cost

```
from agentlab.assistant import ShopAssist2
from prodlab.obs import SAMPLE
from prodlab.perf import answer_cost

bot = ShopAssist2()
for message in [SAMPLE[1], SAMPLE[4], SAMPLE[5]]:
    out = bot.ask("cust-7", message)
    small = answer_cost(message, out, "cloud-small")
    large = answer_cost(message, out, "cloud-large")
    print(f"{message[:38]:<38} calls {out['calls']}")
    tokens = f"{small['tokens_in']}/{small['tokens_out']}"
    print(f"   tokens in/out {tokens}  small {small['credits']:.2f}"
          f"  large {large['credits']:.2f}")
```

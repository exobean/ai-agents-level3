## Deadline

```
from prodlab.perf import FakeClock
from prodlab.reliability import FlakyProvider, call_with_deadline

clock = FakeClock()                      # simulated time: nothing waits
provider = FlakyProvider(clock, seed=4, fail_rate=0.6, slow_rate=0.3)
for request in range(4):
    start = clock.t
    result, log = call_with_deadline(provider, clock, deadline_s=6,
                                     seed=request)
    print(f"request {request}: {result or 'no answer'} "
          f"after {clock.t - start:.1f} s")
    for line in log:
        print("   ", line)
```

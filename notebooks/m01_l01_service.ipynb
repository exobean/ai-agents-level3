## Service

```
from prodlab.service import AgentService, call

with AgentService() as svc:                    # a real HTTP server
    status, body, headers = call(
        svc.url, "/v1/ask",
        {"user": "cust-7", "message": "Where is my order #1042?"},
        headers={"x-request-id": "demo-001"})
    print(status, body["reply"])
    print("request id echoed:", headers["x-request-id"])
    status, body, _ = call(svc.url, "/v1/ask", {"user": "x"})
    print("bad request:", status, body["error"])
```

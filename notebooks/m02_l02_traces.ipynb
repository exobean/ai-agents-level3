## Trace

```
from agentlab.assistant import ShopAssist2
from prodlab.obs import trace_ask

out, spans = trace_ask(ShopAssist2(), "cust-7",
                       "Where is my order #1043? And also, is there "
                       "a student discount?")
root = spans[0]
for span in spans:
    indent = "" if span.parent is None else "  "
    share = 100 * span.ms / root.ms
    print(f"{indent}{span.name:<8} {span.ms:6.0f} ms  {share:3.0f}%")
```

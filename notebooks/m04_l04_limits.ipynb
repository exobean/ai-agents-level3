## Limits

```
from prodlab.perf import FakeClock, RateLimiter

clock = FakeClock()                     # time we control, for the demo
limiter = RateLimiter(rate=0.5, burst=5, clock=clock)
for i in range(8):                      # 8 requests in quick succession
    allowed, retry_after = limiter.allow("heavy-user")
    wait = f"429, retry after {retry_after:.1f} s"
    verdict = "ok" if allowed else wait
    print(f"heavy user, request {i + 1}: {verdict}")
    clock.advance(0.2)
print("other user:", "ok" if limiter.allow("other-user")[0] else "429")
```

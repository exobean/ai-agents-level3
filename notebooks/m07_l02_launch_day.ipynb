## Launch

```
from collections import Counter

from prodlab.launch import day_report, launch_day

rows, bot = launch_day()                # 20 simulated minutes, seeded
report = day_report(rows)
print("requests:", report["requests"])
for outcome, count in sorted(report["outcomes"].items(),
                             key=lambda kv: -kv[1]):
    print(f"   {outcome:<13} {count}")
print("attacks handled safely:", report["attacks_safe"], "of",
      report["attacks"])
print("p95 latency:", report["p95_s"], "s   cost:", report["cost"],
      "credits   review queue:", len(bot.queue.heap))
heavy = Counter(r["outcome"] for r in rows if r["who"] == "heavy")
print("heavy user:")
for outcome, count in heavy.most_common():
    print(f"   {outcome:<13} {count}")
```

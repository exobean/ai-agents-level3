## Output

```
from prodlab.security import SecureAssistant, poisoned_index

guard = SecureAssistant(layers=("output",))     # only the last check
attempts = [
    ("cust-7", "How long does a refund take?"),
    ("cust-8", "Where is order #1042?"),
    ("cust-8", "Change the address of order #1043 to my house."),
]
with poisoned_index():
    for user, message in attempts:
        out = guard.ask(user, message)
        print(f"{user}: {message}")
        print(f"   blocked by: {out['blocked']}")
        print(f"   reply: {out['reply'][:50]}")
```

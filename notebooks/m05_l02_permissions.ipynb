## Permissions

```
from agentlab.assistant import ShopAssist2
from prodlab.security import SecureAssistant

plain = ShopAssist2()
secure = SecureAssistant(layers=("permissions",))
for user, message in [("cust-7", "Where is my order #1042?"),
                      ("cust-8", "Where is order #1042?")]:
    print(f"{user}: {message}")
    print("   plain: ", plain.ask(user, message)["reply"])
    print("   checked:", secure.ask(user, message)["reply"])
```

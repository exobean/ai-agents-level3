## Redteam

```
from agentlab.assistant import ShopAssist2
from prodlab.security import SecureAssistant, run_red_team

for name, make in [("no defences", ShopAssist2),
                   ("all four layers", SecureAssistant)]:
    rows = run_red_team(make)
    safe = sum(r["safe"] for r in rows)
    print(f"{name}: {safe}/{len(rows)} attacks handled safely")
    for r in rows:
        if not r["safe"]:
            print(f"   unsafe: {r['attack']:<16} {r['reply'][:38]}")
```

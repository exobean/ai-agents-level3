## Judge

```
from prodlab.evals import checks, judge

question = "How long does a refund take?"
answers = [
    "Refunds go back to the original payment method within 5 working "
    "days after the parcel reaches our warehouse. [returns#2]",
    "Refunds take time.",
    "Refunds take 7 to 10 days. [returns#2]",
]
for answer in answers:
    failed = [k for k, ok in checks(question, answer).items() if not ok]
    short = answer if len(answer) <= 34 else answer[:34] + "..."
    print(f"score {judge(question, answer)}/4  {short}")
    print("   misses:", failed or "nothing")
```

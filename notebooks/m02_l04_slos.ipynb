## Slo

```
from agentlab.assistant import ShopAssist2
from prodlab.obs import simulate_traffic, slo_report

rows = simulate_traffic(ShopAssist2, n=400)
for target_ms in (1500, 800):
    r = slo_report(rows, target_ms=target_ms, objective=0.95)
    print(f"95% of answers within {target_ms} ms without error:")
    print(f"   achieved {r['compliance']:.0%}, met: {r['met']}, "
          f"error budget used: {r['budget_used']:.0%}")
```
